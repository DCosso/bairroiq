```python

# BarrioIQ - Validacao inicial do ambiente
# Marco 2 - Teste PySpark

from datetime import datetime, timezone

# Cria um pequeno conjunto de dados de teste
dados = [
    ("BarrioIQ", "Desenvolvimento", "OK"),
    ("BarrioIQ", "Engenharia de Dados", "OK"),
    ("BarrioIQ", "DataOps", "OK")
]

# Converte os dados em um DataFrame Spark
df = spark.createDataFrame(
    dados,
    ["projeto", "componente", "status"]
)

# Verifica se o DataFrame foi criado corretamente
assert df.count() == 3, "Quantidade incorreta de registros"

assert (
    df.filter(df.status != "OK").count() == 0
), "Existem registros com status invalido"

# Exibe os resultados
display(df)

print("BarrioIQ: ambiente PySpark validado!")
print("Registros processados:", df.count())
print("Executado em:", datetime.now(timezone.utc))
```
